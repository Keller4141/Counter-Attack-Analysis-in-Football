```python
# Identifies uniqiue events in events.csv
import csv
from collections import Counter
from pathlib import Path


def main() -> None:
    csv_path = Path("Data/events.csv")
    if not csv_path.exists():
        raise FileNotFoundError("Data/events.csv was not found.")

    counts = Counter()
    with csv_path.open(newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        if "typePrimary" not in (reader.fieldnames or []):
            raise ValueError("Column 'typePrimary' is not in events.csv.")
        for row in reader:
            event_type = (row.get("typePrimary") or "").strip()
            if event_type:
                counts[event_type] += 1

    print("Unique events in events.csv:")
    for event_type in sorted(counts):
        print(f"{event_type}: {counts[event_type]}")


if __name__ == "__main__":
    main()
```

```python
# Count number of unique matches in events.csv
import csv

unique_match_ids = set()
with open("Data/events.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    if "matchId" not in (reader.fieldnames or []):
        raise ValueError("Column matchId is not in events.csv.")
    for row in reader:
        match_id = (row.get("matchId") or "").strip()
        if match_id:
            unique_match_ids.add(match_id)

print(f"Antal forskellige kampe: {len(unique_match_ids)}")
```
